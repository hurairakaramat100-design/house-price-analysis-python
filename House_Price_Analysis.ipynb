# House Price Analysis using Python

This beginner-level project explores factors that may affect house prices. The dataset contains area size, number of rooms, location, rating, and age of the house.

**Goal:** Understand the data and identify useful patterns before building a Machine Learning model.

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
```

## 2. Load the Dataset

```python
df = pd.read_csv('house_prices.csv')
df.head()
```

## 3. Understand the Dataset

```python
print('Dataset shape:', df.shape)
df.info()

print('\nSummary statistics:')
df.describe()
```

## 4. Check Missing Values and Duplicates

```python
print('Missing values:')
print(df.isnull().sum())
print('\nDuplicate rows:', df.duplicated().sum())
```

## 5. Basic Price Statistics

```python
print(f"Average house price: {df['House_Price_Lakhs'].mean():.2f} lakhs")
print(f"Minimum house price: {df['House_Price_Lakhs'].min()} lakhs")
print(f"Maximum house price: {df['House_Price_Lakhs'].max()} lakhs")
```

## 6. Area vs House Price

We check whether larger houses tend to have higher prices.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Area_SqFt', y='House_Price_Lakhs', s=70)
plt.title('Area vs House Price')
plt.xlabel('Area (sq ft)')
plt.ylabel('House Price (Lakhs)')
plt.show()
```

## 7. Rooms vs House Price

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Rooms', y='House_Price_Lakhs', s=70)
plt.title('Rooms vs House Price')
plt.xlabel('Number of Rooms')
plt.ylabel('House Price (Lakhs)')
plt.show()
```

## 8. Location vs Average Price

Location is categorical, so we compare the average price of houses in each area.

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Location', y='House_Price_Lakhs', errorbar=None)
plt.title('Average House Price by Location')
plt.xlabel('Location')
plt.ylabel('Average House Price (Lakhs)')
plt.xticks(rotation=20)
plt.show()
```

## 9. House Age vs Price

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Age_Years', y='House_Price_Lakhs', s=70)
plt.title('House Age vs House Price')
plt.xlabel('Age of House (Years)')
plt.ylabel('House Price (Lakhs)')
plt.show()
```

## 10. Correlation Analysis

Correlation helps us understand the relationship between numerical features and house price.

```python
correlation = df[['Area_SqFt', 'Rooms', 'Rating', 'Age_Years', 'House_Price_Lakhs']].corr()
correlation
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

## 12. Main Findings

- Area size has a positive relationship with house price in this sample.
- Houses with more rooms generally have higher prices.
- Location makes a noticeable difference in average price.
